## **Problem Statement**

*   The dataset contains daily trading data for multiple global companies, including fields such as Open, High, Low, Close, Volume, Industry_Tag, Brand_Name, and other relevant stock indicators, categorized by industry and country.

*   The goal is to perform an in-depth stock market analysis using Pandas, focusing on data cleaning, daily return calculations, volatility analysis, industry trends, and correlation between brands and sectors using visualization tools.


*   After implementation, the analysis highlights the most volatile industries, top-performing brands and provides clear visual insights into market trends and inter-stock relationships, demonstrating effective data wrangling and exploratory analysis on real-world stock data.

## **Dataset Info**


**Date:** The date of the stock price data.

**Open:** The opening price of the stock on that date.

**High:** The highest price the stock reached during the trading day.

**Low:** The lowest price the stock reached during the trading day.

**Close:** The closing price of the stock on that date.

**Volume:** The trading volume, i.e., the number of shares traded on that date.

**Brand_Name:** The name of the brand or company.

**Ticker:** Ticker symbol for the stock.

**Industry_Tag:** The industry category or sector to which the brand belongs.

**Country:** The country where the brand is headquartered or primarily operates.

### **Data Type Corrections**
